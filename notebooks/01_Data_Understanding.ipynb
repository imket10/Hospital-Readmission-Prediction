# 📖 Step 10: Target Variable Analysis
---

## 🎯 Objective

The target variable is the output that a machine learning model is trained to predict.

In this project, the target variable is **`readmitted`**, which indicates whether a patient was readmitted to the hospital after discharge.

Before building a machine learning model, it is important to understand the distribution of the target variable.

Analyzing the target variable helps us:

- Identify the different classes present in the dataset.
- Determine whether the dataset is balanced or imbalanced.
- Understand the proportion of patients who were readmitted.
- Select appropriate evaluation metrics and preprocessing techniques.

```python
df["readmitted"].value_counts()
```

## 🔍 Observations

The target variable **`readmitted`** consists of two classes:

- **No:** 13,246 patients were **not readmitted** after discharge.
- **Yes:** 11,754 patients were **readmitted** after discharge.

The difference between the two classes is relatively small, indicating that the dataset is **fairly balanced**.

A balanced dataset is beneficial because the machine learning model can learn patterns from both classes without being heavily biased toward one class.

However, the exact class proportions should also be examined using percentage distribution to confirm the balance.

## 📌 Conclusion

The target variable represents a **binary classification problem**, where the objective is to predict whether a patient will be **readmitted** (`yes`) or **not readmitted** (`no`).

Since both classes are present in comparable numbers, this dataset is well-suited for training classification models without requiring major class balancing techniques.

-------------

## 📊 Percentage Distribution

To better understand the class distribution, we calculate the percentage of each class.

```python
(df["readmitted"].value_counts(normalize=True) * 100).round(2)
```

## 🔍 Observations

The percentage distribution of the target variable is as follows:

- **No:** **52.98%** of patients were **not readmitted** after discharge.
- **Yes:** **47.02%** of patients were **readmitted** after discharge.

The difference between the two classes is relatively small (approximately **6%**), indicating that the dataset is **fairly balanced**.

A balanced dataset is advantageous because the machine learning model has sufficient examples from both classes to learn meaningful patterns. As a result, the model is less likely to become biased toward one particular class.

## 📌 Conclusion

The `readmitted` feature is the **target variable** of this project, making it a **binary classification problem** with two possible outcomes: **Yes** and **No**.

Since the class distribution is nearly balanced (52.98% vs. 47.02%), the dataset is suitable for training classification models without requiring major class-balancing techniques. However, model performance will still be evaluated using appropriate classification metrics such as **Accuracy, Precision, Recall, F1-Score, and ROC-AUC** to ensure reliable predictions.
