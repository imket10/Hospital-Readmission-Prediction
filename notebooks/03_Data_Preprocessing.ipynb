# 📖 Step 5A: Encode the Target Variable
---

## 🎯 Objective

The target variable (`readmitted`) contains categorical values (`Yes` and `No`). Machine learning algorithms require the target variable to be in numerical format before training.

In this step, we use **LabelEncoder** to convert the target labels into numerical values.

This transformation is only applied to the target variable because it represents the class labels that the model must predict.

```python
# ============================================================
# Step 5A: Encode the Target Variable
# ============================================================

from sklearn.preprocessing import LabelEncoder

# Create LabelEncoder object
label_encoder = LabelEncoder()

# Encode target variable
y_encoded = label_encoder.fit_transform(y)

# Display first 10 encoded values
print("First 10 Encoded Values:")
print(y_encoded[:10])

# Display class mapping
print("\nClass Mapping:")
for i, class_name in enumerate(label_encoder.classes_):
    print(f"{class_name} --> {i}")
```

## 🔍 Observations

The target variable has been successfully converted into numerical format using LabelEncoder.

The original class labels have been mapped as follows:

- No → 0
- Yes → 1

This transformation enables machine learning algorithms to interpret the target variable during training.

---
# 📖 Step 5B: One-Hot Encode the Categorical Features
---

## 🎯 Objective

Most machine learning algorithms cannot work directly with categorical (text) data. Therefore, categorical features must be converted into numerical representations.

In this step, we use **One-Hot Encoding**, which creates a separate binary (0 or 1) column for each unique category in a feature.

Unlike Label Encoding, One-Hot Encoding does not introduce any artificial order between categories, making it suitable for nominal categorical variables.

The encoded features will later be combined with the numerical features to create the final dataset for model training.

```python
# ============================================================
# Step 5B: One-Hot Encode the Categorical Features
# ============================================================

from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns
categorical_columns = X.select_dtypes(include=["object", "string"]).columns

# Create OneHotEncoder
encoder = OneHotEncoder(
    sparse_output=False,
    handle_unknown="ignore"
)

# Fit and transform categorical features
X_categorical_encoded = encoder.fit_transform(X[categorical_columns])

# Convert to DataFrame
encoded_df = pd.DataFrame(
    X_categorical_encoded,
    columns=encoder.get_feature_names_out(categorical_columns)
)

# Display shape
print("Encoded Data Shape:", encoded_df.shape)

# Display first five rows
encoded_df.head()
```

## 🔍 Observations

The categorical features have been successfully transformed into numerical format using One-Hot Encoding.

The encoded dataset contains:

- **25,000 observations (rows)**
- **47 encoded categorical features (columns)**

Each unique category has been converted into a separate binary feature containing values of either **0** or **1**.

For example:

- The `age` feature has been expanded into multiple age-group columns.
- The `medical_specialty` feature has been converted into separate specialty-specific columns.
- Binary categorical variables such as `change` and `diabetes_med` have been represented using two binary columns.

This transformation preserves the categorical information while avoiding any artificial ordering between categories.

## 📌 Conclusion

The categorical input features have been successfully encoded into numerical representations using One-Hot Encoding. These encoded features are now suitable for machine learning algorithms.

The next step is to preprocess the numerical features by applying feature scaling before combining both numerical and encoded categorical data.

---
# 📖 Step 5C: Scale the Numerical Features
---

## 🎯 Objective

Numerical features in a dataset often have different value ranges. For example, one feature may range from 1 to 14, while another ranges from 1 to over 100. These differences can negatively affect the performance of many machine learning algorithms.

To overcome this issue, numerical features are standardized using **StandardScaler**.

StandardScaler transforms each numerical feature so that:

- The mean becomes approximately **0**.
- The standard deviation becomes approximately **1**.

This ensures that all numerical features contribute equally during model training and improves the convergence of several machine learning algorithms.

```python
# ============================================================
# Step 5C: Scale Numerical Features
# ============================================================

from sklearn.preprocessing import StandardScaler

# Identify numerical columns
numerical_columns = X.select_dtypes(include=["number"]).columns

# Create StandardScaler object
scaler = StandardScaler()

# Scale numerical features
X_numerical_scaled = scaler.fit_transform(X[numerical_columns])

# Convert NumPy array to DataFrame
scaled_df = pd.DataFrame(
    X_numerical_scaled,
    columns=numerical_columns
)

# Display shape
print("Scaled Numerical Data Shape:", scaled_df.shape)

# Display first five rows
scaled_df.head()
```

## 🔍 Observations

The numerical features have been successfully standardized using **StandardScaler**.

The scaled dataset contains:

- **25,000 observations (rows)**
- **7 numerical features (columns)**

After scaling:

- Each numerical feature has a mean close to **0**.
- Each numerical feature has a standard deviation close to **1**.

The original relationships between observations are preserved while ensuring that all numerical features are on the same scale.

This preprocessing step is particularly beneficial for machine learning algorithms that are sensitive to feature magnitudes, such as Logistic Regression, Support Vector Machines (SVM), K-Nearest Neighbors (KNN), and Neural Networks.

## 📌 Conclusion

The numerical features have been successfully standardized using StandardScaler.

The scaled numerical dataset is now ready to be combined with the encoded categorical features to create the final machine-learning-ready dataset.

---
# 📖 Step 5D: Combine Scaled Numerical and Encoded Categorical Features
---

## 🎯 Objective

After preprocessing, the numerical and categorical features exist as separate datasets.

To prepare the data for machine learning, these transformed features must be combined into a single feature matrix.

The final dataset produced in this step contains:

- Standardized numerical features.
- One-Hot Encoded categorical features.

This combined dataset will serve as the input for machine learning model training.

```python
# ============================================================
# Step 5D: Combine Numerical and Categorical Features
# ============================================================

# Combine numerical and categorical features
X_processed = pd.concat(
    [scaled_df, encoded_df],
    axis=1
)

# Display shape
print("Processed Dataset Shape:", X_processed.shape)

# Display first five rows
X_processed.head()
```

## 🔍 Observations

The scaled numerical features and One-Hot Encoded categorical features have been successfully combined into a single processed dataset.

The final processed dataset contains:

- **25,000 observations (rows)**
- **54 input features (columns)**

The first seven columns represent standardized numerical features, while the remaining forty-seven columns represent One-Hot Encoded categorical features.

All features are now numerical, making the dataset compatible with machine learning algorithms.

The preprocessing steps have preserved the original information while transforming the data into a format suitable for model training.

## 📌 Conclusion

The preprocessing phase has been completed successfully.

The original dataset has been transformed into a machine-learning-ready feature matrix containing standardized numerical features and encoded categorical features.

The processed dataset is now ready to be divided into training and testing subsets for model development.

---
# 📖 Step 6: Split the Dataset into Training and Testing Sets
---

## 🎯 Objective

Before training a machine learning model, the dataset must be divided into two subsets:

- **Training Set** – Used by the model to learn patterns from the data.
- **Testing Set** – Used to evaluate the model on unseen data.

This helps measure how well the model generalizes to new data and prevents overfitting.

In this project:

- **80%** of the data is used for training.
- **20%** of the data is reserved for testing.

The `stratify` parameter is used to preserve the original class distribution in both the training and testing datasets.

```python
# ============================================================
# Step 6: Train-Test Split
# ============================================================

from sklearn.model_selection import train_test_split

# Split the processed dataset
X_train, X_test, y_train, y_test = train_test_split(
    X_processed,
    y_encoded,
    test_size=0.20,
    random_state=42,
    stratify=y_encoded
)

# Display dataset shapes
print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)
print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)
```

## 🔍 Observations

The processed dataset has been successfully divided into training and testing subsets.

The dataset split resulted in:

- **Training Features:** 20,000 observations with 54 processed features.
- **Testing Features:** 5,000 observations with 54 processed features.
- **Training Labels:** 20,000 target values.
- **Testing Labels:** 5,000 target values.

An 80:20 split was used, providing sufficient data for model training while reserving an independent dataset for performance evaluation.

The `stratify` parameter ensured that the distribution of the target classes (`Yes` and `No`) remained approximately the same in both the training and testing datasets.

## 📌 Conclusion

The dataset has been successfully divided into training and testing subsets.

The training dataset will be used to train machine learning models, while the testing dataset will be used to evaluate how well the trained models generalize to unseen patient data.

The dataset is now fully prepared for model development.

```python
processed_df = X_processed.copy()

processed_df["readmitted"] = y_encoded
```
