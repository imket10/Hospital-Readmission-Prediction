# 📖 Step 9: Correlation Analysis
---

## 🎯 Objective

Correlation analysis measures the strength and direction of the relationship between numerical features in the dataset.

A correlation matrix helps identify whether changes in one feature are associated with changes in another feature.

Understanding these relationships is useful because it helps:

- Identify strongly related features.
- Detect potential multicollinearity.
- Understand dependencies between variables.
- Support feature selection before machine learning model development.

In this step, a correlation heatmap is used to visualize the correlation coefficients between all numerical features.

```python
# ============================================================
# Correlation Analysis of Numerical Features
# ============================================================

# Select only numerical columns
numerical_features = df.select_dtypes(include=["int64", "float64"])

# Calculate the correlation matrix
correlation_matrix = numerical_features.corr()

# Create a new figure
plt.figure(figsize=(10, 8))

# Display the correlation matrix as a heatmap
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

# Add a title
plt.title("Correlation Matrix of Numerical Features")

# Display the heatmap
plt.show()
```

## 🔍 Observations

The correlation heatmap illustrates the linear relationships between the numerical features in the dataset. The correlation coefficient ranges from **-1 to +1**, where values closer to **+1** indicate a strong positive relationship, values closer to **-1** indicate a strong negative relationship, and values near **0** indicate little or no linear relationship.

From the heatmap, the following observations can be made:

- **`time_in_hospital`** and **`n_medications`** show the **strongest positive correlation (0.45)** among all numerical features. This suggests that patients who stay longer in the hospital generally receive a greater number of medications.

- **`n_procedures`** and **`n_medications`** have a **moderate positive correlation (0.37)**, indicating that patients undergoing more medical procedures also tend to receive more medications during treatment.

- **`time_in_hospital`** and **`n_lab_procedures`** exhibit a **moderate positive correlation (0.33)**. Patients with longer hospital stays often undergo a larger number of laboratory investigations.

- **`n_lab_procedures`** and **`n_medications`** show a **weak positive correlation (0.27)**, suggesting that patients requiring more laboratory tests generally receive slightly more medications.

- **`n_inpatient`** and **`n_emergency`** have a **weak positive correlation (0.22)**, indicating that patients with previous inpatient admissions may also have a history of emergency department visits.

- Features related to previous healthcare utilization (`n_outpatient`, `n_inpatient`, and `n_emergency`) show **weak correlations** with most other numerical features, suggesting that these variables capture different aspects of patient healthcare history.

- Most remaining feature pairs have correlation values close to **zero**, indicating that there is **no strong linear relationship** between them. This is beneficial because it suggests that the features provide complementary information rather than being highly redundant.

## 📌 Conclusion

The correlation analysis indicates that **no pair of numerical features exhibits a very strong correlation (greater than 0.7)**. This suggests that the dataset has **low multicollinearity**, meaning the numerical features contribute distinct information to the machine learning model.

The strongest relationships are observed between **hospital stay duration, medication count, laboratory procedures, and medical procedures**, which is expected because patients with longer or more complex hospitalizations typically undergo more tests and receive more medications.

Overall, the numerical features appear suitable for predictive modeling, as they provide complementary information without significant redundancy.

# 📖 Step 10: Outlier Detection
---

## 🎯 Objective

Outliers are observations that differ significantly from the majority of the data. They may represent unusual patient cases, measurement errors, or extreme medical conditions.

Detecting outliers helps us:

- Identify unusually high or low values.
- Understand the spread of numerical features.
- Determine whether preprocessing techniques such as outlier treatment or feature scaling are required.
- Improve the quality and reliability of machine learning models.

In this step, box plots are used to visualize the distribution of numerical features and identify potential outliers.

```python
# ============================================================
# Outlier Detection Using Box Plots
# ============================================================

# List of numerical features
numerical_columns = [
    "time_in_hospital",
    "n_lab_procedures",
    "n_procedures",
    "n_medications",
    "n_outpatient",
    "n_inpatient",
    "n_emergency"
]

# Create subplots
fig, axes = plt.subplots(3, 3, figsize=(16, 12))

# Flatten axes for easy iteration
axes = axes.flatten()

# Plot boxplots for each numerical feature
for i, column in enumerate(numerical_columns):
    sns.boxplot(
        data=df,
        y=column,
        ax=axes[i]
    )

    axes[i].set_title(column)

# Remove unused subplot axes
for j in range(len(numerical_columns), len(axes)):
    fig.delaxes(axes[j])

# Adjust spacing
plt.tight_layout()

# Display the plots
plt.show()
```

## 🔍 Observations

The box plots provide a visual representation of the distribution and potential outliers for each numerical feature.

### `time_in_hospital`

- Most patients stayed between **2 and 6 days** in the hospital.
- A few patients remained hospitalized for **13–14 days**, appearing as outliers.
- These longer hospital stays are expected for patients with severe illnesses or complications and therefore represent valid medical observations.

### `n_lab_procedures`

- The majority of patients underwent **30 to 55 laboratory procedures**.
- Several patients required **more than 95 laboratory tests**, resulting in visible outliers.
- These extreme values likely correspond to patients requiring continuous monitoring or extensive diagnostic investigations.

### `n_procedures`

- Most patients underwent **0 to 2 medical procedures** during hospitalization.
- Very few patients required **5 or 6 procedures**, which appear as outliers.
- These cases may represent patients with more complex treatment requirements.

### `n_medications`

- Most patients received **10 to 20 medications**.
- A considerable number of patients received **more than 35 medications**, producing several upper outliers.
- High medication counts may indicate patients with multiple chronic conditions or severe illnesses requiring intensive treatment.

### `n_outpatient`

- The majority of patients had **no previous outpatient visits**.
- A small number of patients had frequent outpatient visits, with some exceeding **30 visits**.
- These observations likely represent patients receiving regular follow-up care for chronic diseases.

### `n_inpatient`

- Most patients had **no previous inpatient admissions**.
- A limited number of patients had multiple prior hospitalizations, with values reaching **15 admissions**.
- These outliers may indicate recurring illnesses or repeated hospital care.

### `n_emergency`

- Most patients had **no previous emergency department visits**.
- A few patients had numerous emergency visits, with the maximum exceeding **60 visits**.
- These extreme observations may represent patients with unstable medical conditions or frequent emergency healthcare utilization.

### Overall Analysis

- Nearly all numerical features contain **upper outliers**, while lower outliers are minimal or absent.
- Most outliers occur in healthcare utilization features (`n_outpatient`, `n_inpatient`, and `n_emergency`) and treatment-related features (`n_medications` and `n_lab_procedures`).
- These outliers are consistent with real-world healthcare data, where a small proportion of patients require significantly more medical attention than the majority.

# 📖 Step 11: Feature vs Readmission Analysis
---

## 🎯 Objective

Understanding how individual features relate to the target variable is an essential part of Exploratory Data Analysis (EDA).

In this step, we compare important patient characteristics with the hospital readmission status to identify potential patterns and relationships.

The objective of this analysis is to:

- Understand how different patient characteristics influence hospital readmission.
- Identify features that may be strong predictors for machine learning models.
- Generate insights that support feature engineering and model selection.
- Discover trends that may be clinically meaningful.

Each visualization compares one feature with the target variable (`readmitted`) to better understand its impact on patient outcomes.

## 🎯 Objective

Age is one of the most important demographic factors in healthcare.

This analysis examines whether hospital readmission varies across different age groups.

Understanding this relationship helps determine whether age is a significant predictor of hospital readmission.

```python
# Readmission Percentage by Age Group

readmission_rate = (
    df.groupby("age")["readmitted"]
      .value_counts(normalize=True)
      .rename("percentage")
      .mul(100)
      .reset_index()
)

plt.figure(figsize=(12, 6))

sns.barplot(
    data=readmission_rate,
    x="age",
    y="percentage",
    hue="readmitted"
)

plt.title("Readmission Percentage by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Percentage (%)")

plt.xticks(rotation=45)
plt.show()
```

## 🔍 Observations

The bar chart presents the percentage of patients who were readmitted (`Yes`) and not readmitted (`No`) within each age group. Unlike the count plot, this visualization provides a fair comparison by accounting for differences in the number of patients across age groups.

From the visualization, the following observations can be made:

- The **80–90 years** age group exhibits the **highest readmission rate**, with approximately **49–50%** of patients being readmitted. This indicates that nearly half of the patients in this age group returned to the hospital after discharge.

- The **70–80 years** age group also shows a relatively high readmission rate of approximately **49%**, suggesting that older patients experience hospital readmissions more frequently.

- Patients aged **60–70 years** have a readmission rate of approximately **47%**, which is slightly lower than the older age groups but still relatively high.

- The **40–50 years** and **50–60 years** age groups have lower readmission rates of approximately **44–45%**, indicating that younger patients are comparatively less likely to be readmitted.

- The **90–100 years** age group shows the **lowest readmission percentage (approximately 42%)**. However, this age group contains the fewest patients in the dataset, so the percentage should be interpreted cautiously.

- Overall, the readmission percentages across age groups range from approximately **42% to 50%**, suggesting that age has a moderate influence on hospital readmission but is not the sole determining factor.

## 📌 Conclusion

The percentage analysis provides a clearer understanding of hospital readmission risk across different age groups than the count-based visualization.

Patients aged **70–90 years** exhibit the highest readmission percentages, indicating that older individuals may have a greater likelihood of returning to the hospital due to age-related health conditions, chronic diseases, or multiple comorbidities.

Although age appears to influence hospital readmission, the relatively small differences between age groups suggest that additional clinical features—such as hospital stay duration, medication count, previous admissions, and laboratory procedures—are also important factors in predicting readmission. Therefore, **age should be retained as a predictive feature** and evaluated alongside other variables during machine learning model development.

# Step 21 - Feature Importance Analysis
---
## Objective

Understanding the importance of individual features helps identify the variables that contribute most to hospital readmission prediction.

Feature importance analysis provides insights into:

- Which patient characteristics have the greatest influence on readmission.
- Which variables contribute the most to predictive performance.
- Whether some features can be removed without significantly affecting the model.

This analysis also supports feature selection before training machine learning models.

```python
# Encode categorical variables

from sklearn.preprocessing import LabelEncoder

df_encoded = df.copy()

label_encoder = LabelEncoder()

for column in df_encoded.select_dtypes(include='object').columns:
    df_encoded[column] = label_encoder.fit_transform(df_encoded[column])

df_encoded.head()
```

```python
# Separate features and target variable

X = df_encoded.drop('readmitted', axis=1)

y = df_encoded['readmitted']
```

```python
# Train Random Forest model

from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf.fit(X, y)

feature_importance = pd.DataFrame({

    "Feature": X.columns,

    "Importance": rf.feature_importances_

})

feature_importance = feature_importance.sort_values(
    by="Importance",
    ascending=False
)

display(feature_importance)
```

```python
# Import the required library
from sklearn.feature_selection import mutual_info_classif

# Calculate Mutual Information scores
mi_scores = mutual_info_classif(
    X,
    y,
    random_state=42
)

# Create a DataFrame
mi = pd.DataFrame({
    "Feature": X.columns,
    "Mutual Information": mi_scores
})

# Sort features by importance
mi = mi.sort_values(
    by="Mutual Information",
    ascending=False
)

display(mi)
```

## Observation

The feature importance analysis identifies the variables that contribute most to predicting hospital readmission using **Random Forest Feature Importance** and **Mutual Information (MI)**. These two techniques provide complementary insights into feature relevance.

### Key Findings

#### Random Forest Feature Importance

- **Number of Laboratory Procedures (`n_lab_procedures`)** is the most influential feature with an importance score of **0.1783**, indicating that laboratory investigations play a significant role in predicting hospital readmission.

- **Number of Medications (`n_medications`)** ranks second (**0.1466**), suggesting that patients receiving more medications are generally more likely to be readmitted due to complex treatment requirements.

- **Time in Hospital (`time_in_hospital`)** is the third most important feature (**0.1001**), reflecting the relationship between prolonged hospitalization and disease severity.

- Patient demographics and diagnosis-related variables, including **Age (`0.0718`)**, **Primary Diagnosis (`diag_1`, 0.0678)**, **Medical Specialty (`0.0673`)**, **Tertiary Diagnosis (`diag_3`, 0.0672)**, and **Secondary Diagnosis (`diag_2`, 0.0655)**, also contribute meaningfully to prediction.

- Hospital utilization variables such as **Previous Inpatient Visits (`0.0547`)**, **Outpatient Visits (`0.0258`)**, and **Emergency Visits (`0.0184`)** provide additional predictive information.

- Diabetes-related variables (**A1Ctest**, **glucose_test**, **change**, and **diabetes_med**) have comparatively lower Random Forest importance scores but still contribute to the predictive model.

#### Mutual Information Analysis

- **Previous Inpatient Visits (`n_inpatient`)** achieved the highest Mutual Information score (**0.0210**), indicating the strongest direct relationship with hospital readmission.

- **Emergency Visits (`0.0086`)**, **Glucose Test (`0.0081`)**, and **Outpatient Visits (`0.0078`)** also show meaningful associations with the target variable.

- Diagnosis-related variables, particularly **Primary Diagnosis (`diag_1`)** and **Secondary Diagnosis (`diag_2`)**, retain useful predictive information.

- Some variables, including **Age**, **Time in Hospital**, **Number of Procedures**, and **Medical Specialty**, received Mutual Information scores close to zero. This suggests that their relationship with the target may be more complex or captured through interactions with other features rather than a simple individual dependency.

### Comparison of Both Methods

- Random Forest emphasizes variables related to **laboratory procedures, medication count, hospital stay duration, and diagnoses**, highlighting their importance in classification.

- Mutual Information places greater emphasis on **previous inpatient admissions, emergency visits, outpatient visits, and glucose test results**, reflecting their direct association with hospital readmission.

- The differences between the two methods indicate that some features contribute through complex interactions captured by Random Forest, while others have stronger individual relationships with the target.

### Clinical Interpretation

The results indicate that **clinical workload (laboratory tests and medications)**, **disease severity (hospital stay duration and diagnoses)**, and **previous healthcare utilization (especially inpatient admissions)** are the most influential factors associated with hospital readmission. Patients requiring intensive investigations, multiple medications, or repeated hospital visits are more likely to experience future readmissions.

### Conclusion

Feature importance analysis confirms that hospital readmission is influenced by a combination of **clinical characteristics, treatment intensity, diagnosis information, and previous healthcare utilization**. Features such as **`n_lab_procedures`**, **`n_medications`**, **`time_in_hospital`**, **`diag_1`**, and **`n_inpatient`** should be retained as key predictors for machine learning models, while lower-ranked features can be evaluated further during feature selection and model optimization.
